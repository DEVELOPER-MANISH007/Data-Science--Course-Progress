```python
## Basic plotting ith seaborn
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x ="total_bill",y = "tip",data = tips)
plt.title("Scatter plot of total bill")
```

## Logging with Multiple Loggers
``` you can create multiple loggers for deifferent parts of your application```

````python
import logging
````

````python
# create a logger module
logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)

## create a logger for moudle2
logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)

# configure logging settings
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s -%(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
````

````python
logger1.debug("this is a debug message for  moudle1")
logger2.warning("this is a warning message for logger2  or moudle 2")
logger2.error("this is a error message for module 2")
````

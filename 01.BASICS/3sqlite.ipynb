```python
import sqlite3
connection = sqlite3.connect("example.db")
```

```python
cursor = connection.cursor()
```

```python
#crete a table
cursor.execute('''
Create Table If  Not exists employees(
    id integer Primary key,
    Name Text Not null,
    age Integer,
    department text
               )
''')

# commit
connection.commit()
```

```python
cursor.execute('''
insert into employees(name, age, department)
values('krish', 32, 'Herosima')
''')

cursor.execute('''
insert into employees(name, age, department)
values('Manish', 12, 'hi ladle')
''')

cursor.execute('''
insert into employees(name, age, department)
values('govind', 43, 'DSA')
''')

connection.commit()
```

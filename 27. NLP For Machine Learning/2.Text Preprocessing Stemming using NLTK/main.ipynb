## ```Stemming```

Stemming is a text preprocessing technique in NLP that reduces different forms of a word to its root word (stem) by removing prefixes or suffixes.

``` 
playing   → play
played    → play
plays     → play
studies   → studi
```

````python
words = [
    "playing",
    "played",
    "plays",
    "playing",
    "running",
    "runs",
    "runner",
    "easily",
    "easy",
    "studies",
    "studying",
    "studied",
    "happily",
    "happiness",
    "connected",
    "connecting",
    "connection",
    "beautiful",
    "beautifully",
    "development"
]
````

## ProterStemmer

````python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

for word in words:
    print(word+"-------->"+stemmer.stem(word))
    

````

### Disadvantages of Stemming

- Can produce incorrect words.
- May remove too much or too little of a word.
- Does not understand word meaning or context.
- Less accurate than Lemmatization.

## ``` RegexpStemmer class```

NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms.

It basically takes a single regular expression and removes any prefix or suffix that matches the expression.



NLTK provides the **RegexpStemmer** class to implement **Regular Expression based stemming**.

It removes the **prefix or suffix** of a word that matches the given **regular expression (regex)**.

````python
from nltk.stem import RegexpStemmer


stemmer = RegexpStemmer('ing$|ed$|ly$|s$|es$|er$|est$|tion$|ment$|ness$', min=4)

print(stemmer.stem("playing"))
````

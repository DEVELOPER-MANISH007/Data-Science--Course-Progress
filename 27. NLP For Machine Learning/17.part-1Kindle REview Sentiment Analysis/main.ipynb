## About Dataset

### Context

This is a small subset of dataset of Book reviews from Amazon Kindle Store category.

### Content

5-core dataset of product reviews from Amazon Kindle Store category from **May 1996 - July 2014**.

Contains a total of **982,619 entries**.

Each reviewer has at least **5 reviews** and each product has at least **5 reviews** in this dataset.

### Columns

- **asin** - ID of the product, like `B000FA64PK`
- **helpful** - Helpfulness rating of the review, example: `2/3`
- **overall** - Rating of the product
- **reviewText** - Text of the review
- **reviewTime** - Time of the review (raw)
- **reviewerID** - ID of the reviewer, like `A3SPTOKDG7WBLN`
- **reviewerName** - Name of the reviewer
- **summary** - Summary of the review
- **unixReviewTime** - Unix timestamp

````python
## preprocessing and Cleaning 
````

````python
## postive review is 1 and neg is 0
df['rating'] = df['rating'].apply(lambda x:0 if x<3 else 1)
````

````python
df['rating'].value_counts()
````

````python
df.head()
````

````python
## 1. Lower All th cases
df['reviewText']=df['reviewText'].str.lower()
````

````python
df.head()
````

````python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
````

````python
from bs4 import BeautifulSoup
````

````python
## Removing special characters
df['reviewText']=df['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
df['reviewText']=df['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
## Remove url 
df['reviewText']=df['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove html tags
df['reviewText']=df['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
## Remove any additional spaces
df['reviewText']=df['reviewText'].apply(lambda x: " ".join(x.split()))
````

````python
df.head()
````

````python
## apply word net Lemmatizer

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
````

````python
def lemmatize_word(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
````

````python
df['reviewText']=df['reviewText'].apply(lambda x:lemmatize_word(x))
````

````python
df.head()
````

````python
## train test split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df['reviewText'],df['rating'] ,test_size=0.2, random_state=42
)
````

#### ``` part to of video has started```

````python
## aplly bag of words

from sklearn.feature_extraction.text import CountVectorizer

bow = CountVectorizer()

X_train_bow = bow.fit_transform(X_train).toarray()
X_test_bow = bow.transform(X_test).toarray()
````

````python

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = CountVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train).toarray()
X_test_tfidf = tfidf.transform(X_test).toarray()
````

````python
X_train_bow
````

````python
from sklearn.naive_bayes import GaussianNB
nb_model_bow = GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf = GaussianNB().fit(X_train_tfidf,y_train)
````

````python
# BOW Prediction
y_pred_bow = nb_model_bow.predict(X_test_bow)

# TF-IDF Prediction
y_pred_tfidf = nb_model_tfidf.predict(X_test_tfidf)
````

````python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# BOW
print("BOW")
print(confusion_matrix(y_test, y_pred_bow))
print(accuracy_score(y_test, y_pred_bow))
print(classification_report(y_test, y_pred_bow))


# TF-IDF
print("\nTF-IDF")
print(confusion_matrix(y_test, y_pred_tfidf))
print(accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf))
````

## Lemmatization

Lemmatization is an NLP technique that reduces a word to its **meaningful dictionary/base form (lemma)** using linguistic rules and context.

### Examples
- `studies → study`
- `running → run`
- `children → child`
- `better → good`

### Key Point
Unlike Stemming, Lemmatization produces a **meaningful and grammatically correct word**.

## Wordnet Lemmatizer

Lemmatization technique is like stemming. The output we will get after lemmatization is called
'lemma', which is a root word rather than root stem, the output of stemming. After lemmatization,
we will be getting a valid word that means the same thing.
NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This
class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us
understand it with an example -

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
words = [
    "playing",
    "played",
    "plays",
    "playing",
    "running",
    "runs",
    "runner",
    "easily",
    "easy",
    "studies",
    "studying",
    "studied",
    "happily",
    "happiness",
    "connected",
    "connecting",
    "connection",
    "beautiful",
    "beautifully",
    "development"
]
```

```python
for word in words:
    print(word+"---------->"+lemmatizer.lemmatize(word,pos='v'))
```

### Disadvantages of Lemmatization

- More computationally expensive than Stemming.
- Slower because it uses linguistic rules and context.
- Requires more resources and processing time.
- May give incorrect results if the context or POS is wrong.

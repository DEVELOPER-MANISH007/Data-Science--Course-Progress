## Part of Speech (POS) Tagging

POS Tagging is the process of assigning a **grammatical category/tag** to each word in a sentence according to its role.

### Examples

- Noun → `book`, `Python`
- Verb → `run`, `learn`
- Adjective → `beautiful`, `good`
- Adverb → `quickly`, `slowly`
- Pronoun → `I`, `he`, `they`

### Common POS Tags

- `NN` → Noun
- `VB` → Verb
- `JJ` → Adjective
- `RB` → Adverb
- `PRP` → Pronoun
- `DT` → Determiner
- `IN` → Preposition
- `CC` → Conjunction

### Example

`I am learning Python.`

`I → PRP`  
`am → VBP`  
`learning → VBG`  
`Python → NNP`

### Key Point

POS Tagging identifies the **grammatical role of each word** in a sentence.

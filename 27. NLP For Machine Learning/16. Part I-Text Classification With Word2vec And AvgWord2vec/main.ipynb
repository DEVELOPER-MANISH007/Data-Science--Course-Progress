```python
import pandas as pd

messages = pd.read_csv('SMSSpamCollection.csv',sep='\t',names=['label','message'])
messages.head()
```

```python
import re
import nltk
nltk.download("stopwords")
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
corpus = []

for i in range(0,len(messages)):
    review= re.sub('^[Aa-zA-z]',' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word)for word  in review]
    review = ' '.join(review)
    corpus.append(review)
    
```

```python
corpus
```

```python
# shortcut for the upper code
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words = []
for sent in corpus:
    sent_token = sent_tokenize(sent)
    words.append(simple_preprocess(sent))
```

```python
import gensim
```

```python
## lets train word2ec from scratch
model = gensim.models.Word2Vec(words)
```

## average Word2Vec

```python
import numpy as np

def avg_word2vec(doc):

    # Remove out-of-vocabulary words
    sent = [word for word in doc if word in model.wv.index_to_key]

    # Agar koi valid word nahi mila
    if len(sent) == 0:
        return np.zeros(model.vector_size)

    return np.mean([model.wv[word] for word in sent], axis=0)
```

```python
from tqdm import tqdm
## apply for the entire sentences
X = []
for i in tqdm(range(len(words))):
    
    X.append(avg_word2vec(words[i]))
```

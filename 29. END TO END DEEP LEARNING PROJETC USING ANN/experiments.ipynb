## Preprocess the data
### Drop irrelevant columns

```python
df = df.drop(['RowNumber','CustomerId','Surname'],axis=1)
```

```python
## Enocde Cateegoorical varibales
lable_encoder_gender = LabelEncoder()
df['Gender'] = lable_encoder_gender.fit_transform(df['Gender'])
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder_geo = OneHotEncoder()
geo_encoder= encoder_geo.fit_transform(df[['Geography']])
```

```python
encoder_geo.get_feature_names_out(['Geography'])
```

```python
geo_encoder.toarray()
```

```python
geo_encoded_df=pd.DataFrame(
    geo_encoder.toarray(),
    columns=encoder_geo.get_feature_names_out(['Geography'])
)
```

```python
## combine one hoot encoder columns with the original data

df = pd.concat([df.drop('Geography',axis=1),geo_encoded_df],axis=1)
df.head()
```

## ANN Implementation

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
## Build Our ANN Model

model =Sequential([
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)),            ## HL1 conncted with input layeer
    Dense(32,activation='relu'), ## HL2
    Dense(1,activation='sigmoid')  ## output layer
])
```

```python
model.summary()
```

```python
import tensorflow
opt =  tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
loss = tensorflow.keras.losses.BinaryCrossentropy()
```

```python
## Compilee the Model
model.compile(optimizer=opt,loss=loss,metrics=['accuracy'])
```

```python
### setup the Tensorboard

from tensorflow.keras.callbacks import EarlyStopping,TensorBoard

log_dir = "logs/fit/"+datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback = TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
## Setup Early Stopping

early_stopping_callback = EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)

```

```python
### Training the Model
history = model.fit(
    X_train,y_train,validation_data = (X_test,y_test),epochs=100,
    
    callbacks=[tensorflow_callback,early_stopping_callback]
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("height-weight.csv")
```

```python
df.head ()
```

```python
plt.scatter(df['Weight'],df['Height'])
plt.xlabel("Weight")
plt.ylabel("Height")
```

```python
## devde our dataset into independent and dependent features
X = df[['Weight']]            ## independent
y = df['Height']            ## dependent feature
```

```python
from sklearn.model_selection  import train_test_split
X_train, X_test,y_train,y_test = train_test_split(X,y,random_state=42, test_size=.20)
```

```python
X_train.shape, X_test.shape,y_train.shape,y_test.shape
```

```python
## Standardize the dataset Train  independent data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
## Train the The Model using Simple Linear Regression
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
```

```python
regressor.fit(X_train,y_train)
```

```python
print("the slope or coffiecent of weight is :",regressor.coef_)
print("Intercept :",regressor.intercept_)
```

```python
plt.scatter(X_train,y_train)
plt.plot(X_train,regressor.predict(X_train),"r")
```

### prediction of train data
1. predicted height output= intercept +coef_(Weights)
2. y_pred_train =157.5 + 17.03(X_train)
          
### prediction of test data
1. predicted height output= intercept +coef_(Weights)
2. y_pred_test =157.5 + 17.03(X_test)

```python
y_pred_test = regressor.predict(X_test)
```

```python
y_pred_test, y_test
```

```python
plt.scatter(X_test,y_test)
plt.plot(X_test,regressor.predict(X_test),'r')
```

### performance Metrics
 - MSE MAE RMSE
-  RSQUARE AND ADJUSTED R SQURE


```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
mse = mean_squared_error(y_test,y_pred_test)
mae = mean_absolute_error(y_test,y_pred_test)
rmse =np.sqrt(mse)
print("MSE:",mse)
print("MAE:",mae)
print("RMSE",rmse)
```

## R square 
Formula

**R^2 = 1 - SSR/SST**


R^2	=	coefficient of determination
SSR	=	sum of squares of residuals
SST	=	total sum of squares

```python
from sklearn.metrics import r2_score
score = r2_score(y_test,y_pred_test)
```

```python
score
```

### Adjusted R2

**Adjusted R2 = 1 – [(1-R2)*(n-1)/(n-k-1)]**

where:

R2: The R2 of the model
n: The number of observations
k: The number of predictor variables

```python
#display adjusted R-squared
1 - (1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

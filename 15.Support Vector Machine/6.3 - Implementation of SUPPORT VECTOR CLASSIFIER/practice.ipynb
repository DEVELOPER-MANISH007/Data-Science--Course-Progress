## Support Vector Classifier Impelemention 

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
## lets create synthetic data point
from sklearn.datasets import make_classification
````

````python
X,y = make_classification(n_samples=1000,n_features=2,n_classes=2,n_redundant=0,n_clusters_per_class=2,random_state=35)
````

````python
X
````

````python
y
````

````python
pd.DataFrame(X)
````

````python
sns.scatterplot(
    x=pd.DataFrame(X)[0],
    y=pd.DataFrame(X)[1],
    hue=y
)

plt.show()
````

````python
## train and split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.25,random_state=10)
````

````python
from sklearn.metrics import confusion_matrix,classification_report
````

### ```BY LINEAR KERNEL```

````python
from sklearn.svm import SVC
linear = SVC(kernel='linear')
````

````python
linear.fit(X_train,y_train)

y_pred = linear.predict(X_test)
````

````python
print("--------------by liner")
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
````

### ```BY RBF KERNEL```

````python
rbf = SVC(kernel='rbf')

rbf.fit(X_train,y_train)
y_pred = rbf.predict(X_test)
````

````python
print("--------------by rbf")
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
````

### ```BY POLYNOMIAL KERNEL```

````python
polynomial = SVC(kernel='poly')


polynomial.fit(X_train,y_train)
y_pred = polynomial.predict(X_test)
````

````python
print("--------------by polynomial-----------------------------")
print("\n")
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
````

## Hyperparametere Tuning With SVC

````python
from sklearn.model_selection import GridSearchCV
 
# defining parameter range
param_grid = {'C': [0.1, 1, 10, 100, 1000],
              'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
              'kernel': ['rbf']}
````

````python
grid=GridSearchCV(SVC(),param_grid=param_grid,refit=True,cv=5,verbose=3)
````

````python
grid.fit(X_train,y_train)
````

````python
grid.best_params_
````

````python
y_pred4 = grid.predict(X_test)
````

````python
print("-------------- Hyperparametere Tuning With SVC ----------------------------")
print("\n")
print(classification_report(y_test,y_pred4))
print(confusion_matrix(y_test,y_pred4))
````
